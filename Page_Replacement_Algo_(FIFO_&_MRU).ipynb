fifo algorithm

```python
referance = [1, 2, 3, 4, 1, 2, 3, 1]
n = 3
faults = 0
mem = []

for p in referance:
  if p not in mem:
    faults += 1
    if len(mem) == n:
      mem.pop(0)
    mem.append(p)

print("fifo page faults :", faults)
```

MRU algorithm


```python
pages = [7, 0, 1, 2, 0, 3, 0, 4,2,3]
capacity = 3
frames = []
last_used = None

for p in pages:

    if p not in frames:

        if len(frames) < capacity:
            frames.append(p)


        else:
            index = frames.index(last_used)
            frames[index] = p

    last_used = p

    print(f"Page: {p} -> Memory: {frames}")
```
